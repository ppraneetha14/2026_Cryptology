```python
def columnar_encrypt(text, key):
    text = "".join(text.split()).upper()
    key = str(key)

    columns = len(key)
    rows = (len(text) + columns - 1) // columns

    text = text.ljust(rows * columns, 'X')

    matrix = [
        text[i:i + columns]
        for i in range(0, len(text), columns)
    ]

    order = sorted(range(columns), key=lambda i: (key[i], i))

    ciphertext = ""
    for col in order:
        for row in matrix:
            ciphertext += row[col]

    return ciphertext


def columnar_decrypt(ciphertext, key):
    key = str(key)
    columns = len(key)
    rows = len(ciphertext) // columns

    order = sorted(range(columns), key=lambda i: (key[i], i))
    matrix = [[''] * columns for _ in range(rows)]

    index = 0
    for col in order:
        for row in range(rows):
            matrix[row][col] = ciphertext[index]
            index += 1

    plaintext = ''.join(''.join(row) for row in matrix)
    return plaintext.rstrip('X')


message = input("Enter plaintext: ")
key = input("Enter numeric key (e.g., 3142): ")

if not key.isdigit() or not key:
    print("Key must contain digits only.")
else:
    encrypted = columnar_encrypt(message, key)
    decrypted = columnar_decrypt(encrypted, key)

    print("Encrypted text:", encrypted)
    print("Decrypted text:", decrypted)
```

```python
def encrypt(text, key):
    cols = len(key)
    rows = (len(text) + cols - 1) // cols

    table = [
        list(text[i:i + cols])
        for i in range(0, len(text), cols)
    ]

    order = sorted(range(cols), key=lambda i: (key[i], i))

    cipher = ""
    for col in order:
        for row in table:
            if col < len(row):
                cipher += row[col]

    return cipher


def decrypt(cipher, key):
    cols = len(key)
    length = len(cipher)
    rows = (length + cols - 1) // cols

    table = [[''] * cols for _ in range(rows)]
    order = sorted(range(cols), key=lambda i: (key[i], i))

    short_cols = rows * cols - length
    pos = 0

    for col in order:
        col_length = rows - 1 if col >= cols - short_cols else rows

        for row in range(col_length):
            table[row][col] = cipher[pos]
            pos += 1

    plain = ""
    for row in table:
        for ch in row:
            plain += ch

    return plain


text = input("Enter plaintext: ").replace(" ", "").upper()
key1 = input("Enter first word key: ").upper()
key2 = input("Enter second word key: ").upper()

if not key1 or not key2:
    print("Keys cannot be empty.")
else:
    c1 = encrypt(text, key1)
    c2 = encrypt(c1, key2)

    print("\nFirst encryption:", c1)
    print("Final ciphertext:", c2)

    d1 = decrypt(c2, key2)
    d2 = decrypt(d1, key1)

    print("\nFirst decryption:", d1)
    print("Final plaintext:", d2)
```
