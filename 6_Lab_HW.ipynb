```python
#One Time Pad
plaintext = input("Enter plaintext: ").upper()
key = input("Enter key: ").upper()

if len(plaintext) != len(key):
    print("Key length must be same as plaintext length.")
else:
    ciphertext = ""

    for i in range(len(plaintext)):
        p = ord(plaintext[i]) - ord('A')
        k = ord(key[i]) - ord('A')
        c = (p + k) % 26
        ciphertext += chr(c + ord('A'))

    print("Ciphertext:", ciphertext)

    decrypted = ""

    for i in range(len(ciphertext)):
        c = ord(ciphertext[i]) - ord('A')
        k = ord(key[i]) - ord('A')
        p = (c - k) % 26
        decrypted += chr(p + ord('A'))

    print("Decrypted text:", decrypted)
```

```python
#shift cipher "MCI RWR WH" (Perform a brute-force attack by displaying all 26 possible plaintexts.)
ciphertext = "MCI RWR WH"

for shift in range(26):
    plaintext = ""

    for ch in ciphertext:
        if ch.isalpha():
            plaintext += chr((ord(ch) - ord('A') - shift) % 26 + ord('A'))
        else:
            plaintext += ch

    print("Shift", shift, ":", plaintext)
```

```python
#Encrypt Rail Fence
text = input("Enter plaintext: ").replace(" ", "")
rails = int(input("Enter number of rails: "))

rail = ['' for i in range(rails)]

row = 0
direction = 1

for ch in text:
    rail[row] += ch

    if row == 0:
        direction = 1
    elif row == rails - 1:
        direction = -1

    row += direction

ciphertext = ""

for r in rail:
    ciphertext += r

print("Ciphertext:", ciphertext)
```
