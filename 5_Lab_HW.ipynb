```python
#Encryption and Decryption (SHIFT CIPHER)
def shift_encrypt(text, key):
    result = ""

    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            result += chr((ord(ch) - base + key) % 26 + base)
        else:
            result += ch

    return result


def shift_decrypt(text, key):
    return shift_encrypt(text, -key)


text = input("Enter text: ")

try:
    key = int(input("Enter shift value: "))

    if key < 0:
        print("Shift value must be non-negative.")
    else:
        print("Encryption:", shift_encrypt(text, key))
        print("Decryption:", shift_decrypt(shift_encrypt(text, key), key))

except ValueError:
    print("Invalid shift value. Enter an integer.")
```

```python
#Encryption and Decryption (CAESAR CIPHER)
def caesar_encrypt(text):
    result = ""

    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            result += chr((ord(ch) - base + 3) % 26 + base)
        else:
            result += ch

    return result


def caesar_decrypt(text):
    result = ""

    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            result += chr((ord(ch) - base - 3) % 26 + base)
        else:
            result += ch

    return result


text = input("Enter text: ")

if text.strip() == "":
    print("Text cannot be empty.")
else:
    encrypted = caesar_encrypt(text)
    print("Encrypted:", encrypted)
    print("Decrypted:", caesar_decrypt(encrypted))
```

```python
#Encryption and Decryption (ROT 13)
def rot13(text):
    result = ""

    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            result += chr((ord(ch) - base + 13) % 26 + base)
        else:
            result += ch

    return result


text = input("Enter text: ")

if text.strip() == "":
    print("Text cannot be empty.")
else:
    encrypted = rot13(text)
    print("Encrypted:", encrypted)
    print("Decrypted:", rot13(encrypted))
```

```python
#Encryption and Decryption (Atbash Cipher)
def atbash(text):
    result = ""

    for ch in text:
        if ch.isupper():
            result += chr(ord('Z') - (ord(ch) - ord('A')))
        elif ch.islower():
            result += chr(ord('z') - (ord(ch) - ord('a')))
        else:
            result += ch

    return result


text = input("Enter text: ")

if text.strip() == "":
    print("Text cannot be empty.")
else:
    encrypted = atbash(text)
    print("Encrypted:", encrypted)
    print("Decrypted:", atbash(encrypted))
```

```python
#Encryption and Decryption (AFFINE CIPHER)
from math import gcd

def affine_encrypt(text, a, b):
    result = ""

    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            x = ord(ch) - base
            result += chr((a * x + b) % 26 + base)
        else:
            result += ch

    return result

def affine_decrypt(text, a, b):
    result = ""

    a_inverse = pow(a, -1, 26)

    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            y = ord(ch) - base
            result += chr((a_inverse * (y - b)) % 26 + base)
        else:
            result += ch

    return result

text = input("Enter text: ")

try:
    a = int(input("Enter value of a: "))
    b = int(input("Enter value of b (0-25): "))

    if gcd(a, 26) != 1:
        print("Invalid value of a. gcd(a, 26) must be 1.")
    elif b < 0 or b > 25:
        print("Invalid value of b. It must be between 0 and 25.")
    else:
        encrypted = affine_encrypt(text, a, b)
        print("Encrypted:", encrypted)
        print("Decrypted:", affine_decrypt(encrypted, a, b))

except ValueError:
    print("Invalid input. Enter integers only.")
```
